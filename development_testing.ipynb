```python
%sql

SELECT 'customer' AS table_name, COUNT(*) AS row_count
FROM nugabanktransactions.gold.customer

UNION ALL

SELECT 'employee', COUNT(*)
FROM nugabanktransactions.gold.employee

UNION ALL

SELECT 'transaction', COUNT(*)
FROM nugabanktransactions.gold.transaction

UNION ALL

SELECT 'fact_table', COUNT(*)
FROM nugabanktransactions.gold.fact_table;
```

```python
%sql

SELECT
    SUM(CASE WHEN customer_id IS NULL THEN 1 ELSE 0 END) AS missing_customer_ids,
    SUM(CASE WHEN employee_id IS NULL THEN 1 ELSE 0 END) AS missing_employee_ids,
    SUM(CASE WHEN transaction_id IS NULL THEN 1 ELSE 0 END) AS missing_transaction_ids
FROM nugabanktransactions.gold.fact_table;
```
